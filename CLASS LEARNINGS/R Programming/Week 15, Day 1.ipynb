```r
# Plot Labels: The plot() function also accepts other parameters, such as main, xlab and ylab if you want to customize the graph with a main title and different labels for the x and y-axis:

# Example
plot(1:10, main="My Graph", xlab="The x-axis", ylab="The y axis")
```

```r
# Graph Appearance: There are many other parameters you can use to change the appearance of the points.
# Colours: use col="color" to add a color to the points:
# Example
plot(1:10, col="red")
```

```r
# Size: use cex=number to change the size of the points (1 is default, while 0.5 means 50% smaller, and 2 means 100% larger)
# cex means character expansion. It controls the size of points, text, or symbols in a plot.
# Example
plot(1:10, cex=0.5) # cex<1 will be smaller
plot(1:10, cex=1) # cex=1 is the normal size or default size
plot(1:10, cex=2) # cex>1 will be bigger
```

```r
# character e
plot(1:10, main="My Plot", xlab="The x-axis", ylab="The y axis", cex.main=2, cex.lab=2)
```

```r
plot(1:10,
     main="My Plot",
     xlab="",
     ylab="",
     cex.main=4)

# Add custom X-axis label
mtext("The x-axis", side = 1, line = 2, cex = 1.5)

# Add custom Y-axis label
mtext("The y-axis", side = 2, line = 2, cex = 3)
```

```r
# Point Shape: Use pch with a value from 0 to 25 to change the point shape format. The values of the pch parameter ranges from 0 to 25, whcih means that we can choose up to 26 different types of point shapes.
# Example
plot(1:10, pch=25, cex=2)
plot(1:10, pch=22, cex=2)
```

R Line: A line graph has a line that connects all points in adiagram. To create a line, use the plot() function and add the type parameter with a value of "l".

```r
# Example
plot(1:10, type="l")
```

```r
# Line Width: To change the width of the line, use the lwd parameter (1 is default, while 0.5 means 50% smaller, and 2 means 100% larger)
# Example
plot(1:10, type="l",lwd=2)
```

```r
# Line Styles: The line is solid by default. Use the lty parameter with a value from 0 to 6 to specify the line format.
# For example, lty=3 will display a dotted line instead of a solid line
# Example
plot(1:10, type="l", lwd=5, lty=3)
```

```r
# Multiple lines: To display more than one line in a graph, use the plot() function together with the lines() function:
# Example:
line1 <- c(1,2,3,4,5,10)
line2 <- c(2,5,7,8,9,10)

plot(line1, type = "l", col = "blue") # This creates the line graph
lines(line2, type="l", col = "red") # It does not create a new graph instead, it adds to the current one
# Note: if you use plot() twice, it will overwrite the first graph
# Note: thats why lines() is used for multiple lines.
```

```r
plot(1:10,
     type = "l",
     col ="blue", # col = line color
     lwd = 2, # lwd - line thickness
     main = "Line Plot Example")
```
