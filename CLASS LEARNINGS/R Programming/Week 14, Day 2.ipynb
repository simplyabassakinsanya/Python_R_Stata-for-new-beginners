# R Factors

## Factors

A factor is a data structure in R used to represent categorical data (groups or labels).

Instead of treating values as plain text, R treats them as defined categories with meaning.

Factors are used to categorize data. Examples of factors are:

- Demography: Male/Female

- Music: Rock, Pop, Classic, Jazz

- Training: Strength, Stamina

**To create a factor, use the factor() function and add a vector as argument:**

```r
# Without factors, R sees this just text (characterf data)
c("Male", "Female", "Male")
```

```r
# R now understands these are categories with possible repetition and structure.
factor(c("Male", "Female", "Male"))
```

Importance/Key reasons for R Factor

- Factor are used to categorized data types properly.
- It also enables us to group and aggregate it.
- It helps to improve statistical analysis.
- To ensure orderliness

## Key Reasons We Use Factors

1. To Represent Categorical Data Properly

Real-world data is full of categories:

- Gender

- Product Type

- Region

- Training Type

**Factors help R treat them correctly**

2. To Enable Grouping & Aggregation

3. To Improve Statistical Analysis

4. To control order (very important)

```r
factor(c("Low", "Medium", "High"),
       levels = c("Low", "Medium", "High"),
       ordered = TRUE)


factor(c("Low", "Medium", "High"),
       levels = c("Low", "Medium", "High"),
       ordered = FALSE)
```

```r
# Example
# Create a factor
music_genre <- factor(c("Jazz", "Rock", "Classic", "Classic", "Pop", "Jazz", "Rock", "Jazz"))

# Print the factor
music_genre
```

```r
# Create a simple character vector
training <- c("Strength", "Stamina", "Strength", "Other")

# convert to factor
factor_training <- factor(training)

# print result
factor_training
```

```r
# You can see from the example above that the factor has four levels(categories): classic, jazz, pop,. and rock
# To only print the levels, use the levels() function:
# Example

music_genre <- factor(c("Jazz", "Rock", "Classic", "Classic", "Pop", "Jazz", "Rock", "Jazz"))
levels(music_genre)
```

```r
# You can also set the levels, by adding the levels argument inside the factor() function:
music_genre <- factor(c("Jazz", "Rock", "Classic", "Classic", "Pop", "Jazz", "Rock", "Jazz"), levels = c("Classic", "Jazz", "Pop", "Rock", "Other"))
levels(music_genre)
```

```r
# Length of a factor
music_genre <- factor(c("Jazz", "Rock", "Classic", "Classic", "Pop", "Jazz", "Rock", "Jazz"))
length(music_genre)
```

```r
# Access factors: To access the items in a factor, refer to the index number, using [] brackets:
music_genre <- factor(c("Jazz", "Rock", "Classic", "Classic", "Pop", "Jazz", "Rock", "Jazz"))
music_genre[3]
```

```r
# Change item value: To change the value of the third item.
music_genre <- factor(c("Jazz", "Rock", "Classic", "Classic", "Pop", "Jazz", "Rock", "Jazz"))
music_genre[3] <- "Pop"
music_genre[3]
music_genre
```

Note that you cannot change the value of a specific item if it is not already specified in the factor. The following example will produce an error:

Example

Trying to change the value of the third item ("Classic") to an item that does not exist/not predefined ("Opera"):

```r
music_genre <- factor(c("Jazz", "Rock", "Classic", "Classic", "Pop", "Jazz", "Rock", "Jazz"))

music_genre[3] <- "Opera"

music_genre[3]
```
