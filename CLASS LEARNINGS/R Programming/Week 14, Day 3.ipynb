# R Plotting
# Plot()

# This is a built-in R function used to create graphs.

By default, it creates a scatter plot (points on a cordinate plane).

The plot() function is used to draw points (markers) in a diagram.

The function takes parameters for specifying points in the diagram

Parameter 1 specifies points on the x-axis

Parameter 2 speccifies points on the y-axis.

At its simplest, you can use the plot() function to plot two numbers against each other.

```r
# Example
# Draw one point in the diagram, at position (1) and position (3)

plot(1, 3)
```

```r
# To draw more points, use vectors:

# Example
# Draw two points in the diagram, one at position (1, 3) and one in position (8, 10):

plot(c(1, 8), c(3, 10)) # c() means combine (or create a vector).
                        # c(1, 8) - X-axis values i.e creates a vector of X cordinates.
                        # c(3, 10) - Y-axis values i.e creates a vector of Y cordinates
```

Visual Understanding

Think of it like:

Point 1 → (1, 3)
Point 2 → (8, 10)

R automatically maps:

- First X → First Y
- Second X → Second Y

In conclusion, R connects values by position - the first X goes with the first Y, the second X with the second Y.

```r
install.packages("scatterplot3d")
library(scatterplot3d)
x <- c(1, 8, 5)
y <- c(3, 10, 7)
z <- c(4, 12, 9)

scatterplot3d(x, y, z, main ="3D Scatter Plot")
```

```r
# Example
plot(c(1, 2, 3, 4, 5), c(3, 7, 8, 9, 12))
```

```r
# For better organization, when you have many values, it is better to use variables.

# Example
x <- c(1, 2, 3, 4, 5)
y <- c(3, 7, 8, 9, 12)

plot(x, y)
```

```r
# Example
plot(1:10)          # 1:10 → Sequence Operator (:)
                    # The : operator in R creates a sequence of numbers.
                    # plot(1:10) is the same as writing: plot(1:10, 1:10)
                    # So it draws points along a straight diagonal line from: (1,1) → (10,10)
```

```r
# Example 2: Reverse order
# Creates a descending line

plot(10:1)
```

```r
plot(5:15)
```

When you write plot(5:15), R sees only one set of values, so it automatically uses the position of each value (1, 2, 3, …) as the x-axis and the actual numbers (5, 6, 7, …, 15) as the y-axis.
In other words, R is not using 5 to 15 as x-values; instead, it counts how many values are present (11 values) and assigns x-values from 1 to 11, pairing each position with its corresponding value.

```r
# Add Style

plot(1:10, type = "b", col = "blue")
plot(1:10, type = "b", col = "black")

# type = "b" - both points and lines.
# col = "blue" - colour.
```

```r
# Making it Clearer Visually
plot(5:15,
     main = "Plot of 5:15",
     xlab = "Index (X)",
     ylab = "Values (Y)",
     col = "red",
     type = "b")
```

```r
# Draw a line: The plot() function also takes a type parameter with the value l to draw a line to connect all the points in the diagram:
# Example

plot(1:10, type="l")
plot(1:10, type="s")
```

### R plot() Type Options
| Type  | Name                   | Description                                | What It Looks Like |   |   |   |
| ----- | ---------------------- | ------------------------------------------ | ------------------ | - | - | - |
| `"p"` | Points (default)       | Plots only points (dots)                   | • • • •            |   |   |   |
| `"l"` | Lines                  | Connects points with lines                 | ─────              |   |   |   |
| `"b"` | Both                   | Points + lines (points visible)            | •─•─•              |   |   |   |
| `"o"` | Overplotted            | Points and lines overlap exactly           | ●──●               |   |   |   |
| `"c"` | Lines only (no points) | Like `"b"` but without points              | ─────              |   |   |   |
| `"s"` | Steps                  | Stair-step plot (horizontal then vertical) | ⎾⎿⎾                |   |   |   |
| `"S"` | Steps (reverse)        | Vertical then horizontal steps             | ⎿⎾⎿                |   |   |   |
| `"h"` | Histogram-like         | Vertical lines from x-axis to points       |                    |   |   |   |
| `"n"` | None                   | Plots nothing (just axes)                  | Empty plot         |   |   |   |

This code below creates a list of different plot styles, then uses a loop to display the same data multiple times, each time using a different style, while pausing briefly between each plot so you can clearly see the differences.                                                         

```r
# This shows all types one after another visually
types <- c("p","l","b","o","c","s","S","h","n")

for (t in types) {
  plot(1:10, type = t, main = paste("type =", t))
  Sys.sleep(1)
}
```
